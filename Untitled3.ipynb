```python

def sample_uniform(a, b):
    u = random.random()
    x = a + (b - a) * u
    return x

# Test the function
print(sample_uniform(1, 5))
print(sample_uniform(1, 5))
print(sample_uniform(1, 5))
```

```python

def sample_normal(mu, sigma):
    u1 = 1 - random.random()
    u2 = random.random()

    z = math.sqrt(-2 * math.log(u1)) * math.cos(2 * math.pi * u2)

    x = mu + sigma * z
    return x

# Test the function
print(sample_normal(3, 1))
print(sample_normal(3, 1))
print(sample_normal(3, 1))
```

```python

def sample_poisson(lam):
    L = math.exp(-lam)
    k = 0
    p = 1

    while True:
        u = random.random()
        p = p * u

        if p <= L:
            return k
        else:
            k = k + 1

# Test the function
print(sample_poisson(2))
print(sample_poisson(2))
print(sample_poisson(2))
```

```python

# Generate 100,000 samples from each distribution

N = 100_000

uniform_samples = [sample_uniform(0, 1) for _ in range(N)]
normal_samples = [sample_normal(3, 1) for _ in range(N)]
poisson_samples = [sample_poisson(2) for _ in range(N)]


# Calculate mean and population variance
def calculate_mean(data):
    return sum(data) / len(data)


def calculate_variance(data):
    mean = calculate_mean(data)
    total = 0

    for x in data:
        total += (x - mean) ** 2

    return total / len(data)


# Print results
for name, data in [
    ("Uniform", uniform_samples),
    ("Normal", normal_samples),
    ("Poisson", poisson_samples)
]:
    print(f"\n{name} Distribution")
    print("Mean:", round(calculate_mean(data), 4))
    print("Variance:", round(calculate_variance(data), 4))

```

```python

def generate_ratings(n, dist, **params):
    ratings = []

    for _ in range(n):

        if dist == "uniform":
            x = sample_uniform(0.5, 5.5)

        elif dist == "normal":
            mu = params.get("mu", 3)
            sigma = params.get("sigma", 1)
            x = sample_normal(mu, sigma)

        elif dist == "poisson":
            lam = params.get("lam", 2)
            x = 1 + sample_poisson(lam)

        else:
            raise ValueError("Unknown distribution")

        rating = round(x)
        rating = max(1, min(5, rating))

        ratings.append(rating)

    return ratings


# Test the function
print("Uniform:", generate_ratings(10, "uniform"))
print("Normal:", generate_ratings(10, "normal"))
print("Poisson:", generate_ratings(10, "poisson"))
```

```python

def assign_ratings(n_items, dist, max_fill=0.5, **params):
    # All items are initially missing
    row = [None] * n_items

    # Choose how many items the user will rate: 5 to 50
    max_rated = int(max_fill * n_items)
    k = random.randint(5, max_rated)

    # Select distinct item positions
    chosen_items = random.sample(range(n_items), k)

    # Generate ratings for the selected items
    ratings = generate_ratings(k, dist, **params)

    # Place ratings in their selected positions
    for item_index, rating in zip(chosen_items, ratings):
        row[item_index] = rating

    # Check that at least 50% of items are missing
    assert row.count(None) >= n_items * 0.5

    return row


# Test with one user and 100 items
random.seed(425)

user_row = assign_ratings(100, "uniform")

print("Total items:", len(user_row))
print("Rated items:", sum(x is not None for x in user_row))
print("Missing items:", user_row.count(None))
print("User ratings:", user_row)

```

```python

import csv

def save_matrix_to_csv(matrix, filename):
    with open(filename, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["user_id", "item_id", "rating"])

        for user_index, row in enumerate(matrix):
            for item_index, rating in enumerate(row):
                if rating is not None:
                    writer.writerow([
                        user_index,
                        item_index,
                        rating
                    ])

    print(f"Saved: {filename}")

save_matrix_to_csv(uniform_matrix, "uniform_ratings.csv")
save_matrix_to_csv(normal_matrix, "normal_ratings.csv")
save_matrix_to_csv(poisson_matrix, "poisson_ratings.csv")
```

```python

def analyze_matrix(matrix, name):
    all_ratings = []
    user_means = []
    item_means = []

    # Collect all observed ratings
    for row in matrix:
        observed = [r for r in row if r is not None]
        all_ratings.extend(observed)

        if observed:
            user_means.append(sum(observed) / len(observed))

    # Calculate each item's mean
    for j in range(len(matrix[0])):
        item_ratings = [
            row[j] for row in matrix
            if row[j] is not None
        ]
        if item_ratings:
            item_means.append(sum(item_ratings) / len(item_ratings))

    total_cells = len(matrix) * len(matrix[0])
    observed_count = len(all_ratings)
    sparsity = 1 - observed_count / total_cells

    global_mean = sum(all_ratings) / len(all_ratings)
    mean_of_user_means = sum(user_means) / len(user_means)

    print(f"\n{name} Matrix")
    print("-" * 35)
    print("Global mean:", round(global_mean, 4))
    print("Mean of user means:", round(mean_of_user_means, 4))
    print("Minimum user mean:", round(min(user_means), 4))
    print("Maximum user mean:", round(max(user_means), 4))
    print("Minimum item mean:", round(min(item_means), 4))
    print("Maximum item mean:", round(max(item_means), 4))
    print("Sparsity:", round(sparsity * 100, 2), "%")

    return all_ratings, user_means, item_means


uniform_stats = analyze_matrix(uniform_matrix, "Uniform")
normal_stats = analyze_matrix(normal_matrix, "Normal")
poisson_stats = analyze_matrix(poisson_matrix, "Poisson")
```

```python

import math
import matplotlib.pyplot as plt

def calculate_homogeneity(all_ratings, name):
    total = len(all_ratings)

    # Count ratings 1, 2, 3, 4, 5
    counts = [
        all_ratings.count(rating)
        for rating in range(1, 6)
    ]

    percentages = [
        count / total * 100
        for count in counts
    ]

    # Normalized entropy
    entropy = 0
    for count in counts:
        if count > 0:
            p = count / total
            entropy -= p * math.log(p)

    normalized_entropy = entropy / math.log(5)

    # Chi-square against an equal distribution
    expected = total / 5
    chi_square = sum(
        (count - expected) ** 2 / expected
        for count in counts
    )

    print(f"\n{name}")
    print("Rating counts (1 to 5):", counts)
    print("Rating percentages (%):",
          [round(p, 2) for p in percentages])
    print("Normalized entropy:", round(normalized_entropy, 4))
    print("Chi-square:", round(chi_square, 2))

    return counts, percentages, normalized_entropy, chi_square


uniform_homogeneity = calculate_homogeneity(
    uniform_stats[0], "Uniform"
)

normal_homogeneity = calculate_homogeneity(
    normal_stats[0], "Normal"
)

poisson_homogeneity = calculate_homogeneity(
    poisson_stats[0], "Poisson"
)
```

```python

def sample_uniform(a, b):
    u = random.random()
    return a + (b - a) * u


def sample_normal(mu, sigma):
    u1 = 1 - random.random()
    u2 = random.random()
    z = math.sqrt(-2 * math.log(u1)) * math.cos(2 * math.pi * u2)
    return mu + sigma * z


def sample_poisson(lam):
    L = math.exp(-lam)
    k = 0
    p = 1

    while True:
        u = random.random()
        p *= u

        if p <= L:
            return k
        k += 1


def generate_ratings(n, dist, **params):
    ratings = []

    for _ in range(n):
        if dist == "uniform":
            x = sample_uniform(0.5, 5.5)

        elif dist == "normal":
            mu = params.get("mu", 3)
            sigma = params.get("sigma", 1)
            x = sample_normal(mu, sigma)

        elif dist == "poisson":
            lam = params.get("lam", 2)
            x = 1 + sample_poisson(lam)

        else:
            raise ValueError("Unknown distribution")

        rating = round(x)
        rating = max(1, min(5, rating))
        ratings.append(rating)

    return ratings


print("All sampler and rating functions are ready!")
```

```python

def hide_ratings(matrix, p, mechanism="random"):
    # Make a copy so the original matrix stays unchanged
    hidden_matrix = [row.copy() for row in matrix]

    n_users = len(matrix)
    n_items = len(matrix[0])
    total_cells = n_users * n_items

    if not 0 <= p <= 1:
        raise ValueError("p must be between 0 and 1")

    if mechanism == "random":
        # Select the cells to hide randomly
        all_positions = [
            (u, i)
            for u in range(n_users)
            for i in range(n_items)
        ]

        n_hide = round(p * total_cells)
        hidden_positions = random.sample(all_positions, n_hide)

        for u, i in hidden_positions:
            hidden_matrix[u][i] = None

    elif mechanism == "dislike":
        # Low ratings are less likely to remain visible
        keys = []

        for u in range(n_users):
            for i in range(n_items):
                rating = matrix[u][i]
                r = max(1, rating)
                rand_value = max(random.random(), 1e-12)
                key = rand_value ** (1 / r)
                keys.append((key, u, i))

        n_keep = round((1 - p) * total_cells)
        keys.sort(reverse=True)

        keep_positions = {
            (u, i) for key, u, i in keys[:n_keep]
        }

        for u in range(n_users):
            for i in range(n_items):
                if (u, i) not in keep_positions:
                    hidden_matrix[u][i] = None

    else:
        raise ValueError("Unknown mechanism")

    return hidden_matrix


# Quick test: hide 50% of ratings randomly
random.seed(425)
test_hidden = hide_ratings(full_matrix, 0.50, "random")

remaining = sum(
    rating is not None
    for row in test_hidden
    for rating in row
)

print("Total ratings:", N_USERS * N_ITEMS)
print("Remaining ratings:", remaining)
print("Hidden percentage:",
      round((1 - remaining / (N_USERS * N_ITEMS)) * 100, 2), "%")
```

```python

def run_missing_data_experiment_fast():
    results = []
    n_users = len(full_matrix)
    n_items = len(full_matrix[0])
    total_cells = n_users * n_items

    # Flatten the matrix into one list
    flat_ratings = [
        rating
        for row in full_matrix
        for rating in row
    ]

    hide_percentages = [0.50, 0.75, 0.90]

    for mechanism in ["random", "dislike"]:
        trial_diffs = {
            p: {"global": [], "user": []}
            for p in hide_percentages
        }

        for trial in range(20):
            random.seed(425 + trial)

            if mechanism == "random":
                # A random ordering of all cell positions
                order = random.sample(
                    range(total_cells), total_cells
                )

            else:
                # Rank cells by dislike-biased retention key
                keyed_positions = []

                for idx, rating in enumerate(flat_ratings):
                    u = max(random.random(), 1e-12)
                    key = u ** (1 / max(1, rating))
                    keyed_positions.append((key, idx))

                keyed_positions.sort(reverse=True)
                order = [idx for key, idx in keyed_positions]

            # Reuse this ordering for all three hiding percentages
            for p in hide_percentages:
                n_keep = round((1 - p) * total_cells)
                kept_indices = order[:n_keep]

                total_rating = 0
                user_sums = [0] * n_users
                user_counts = [0] * n_users

                for idx in kept_indices:
                    rating = flat_ratings[idx]
                    user_idx = idx // n_items

                    total_rating += rating
                    user_sums[user_idx] += rating
                    user_counts[user_idx] += 1

                global_mean = total_rating / n_keep

                observed_user_means = [
                    user_sums[u] / user_counts[u]
                    for u in range(n_users)
                    if user_counts[u] > 0
                ]

                mean_user_means = (
                    sum(observed_user_means)
                    / len(observed_user_means)
                )

                trial_diffs[p]["global"].append(
                    global_mean - reference_global_mean
                )
                trial_diffs[p]["user"].append(
                    mean_user_means - reference_mean_of_user_means
                )

            print(
                f"\r{mechanism}: completed trial {trial + 1}/20",
                end=""
            )

        print()

        for p in hide_percentages:
            global_diffs = trial_diffs[p]["global"]
            user_diffs = trial_diffs[p]["user"]

            results.append({
                "Mechanism": mechanism,
                "Hidden": f"{int(p * 100)}%",
                "Global mean bias": sum(global_diffs) / 20,
                "Global mean error": population_std(global_diffs),
                "User means bias": sum(user_diffs) / 20,
                "User means error": population_std(user_diffs)
            })

    return results


experiment_results = run_missing_data_experiment_fast()

for result in experiment_results:
    print("\n", result["Mechanism"], "| Hidden:", result["Hidden"])
    print("Global mean bias:",
          round(result["Global mean bias"], 4))
    print("Global mean random error:",
          round(result["Global mean error"], 4))
    print("Mean of user means bias:",
          round(result["User means bias"], 4))
    print("Mean of user means random error:",
          round(result["User means error"], 4))
```

```python

import random

def add_cold_start(matrix, dist):
    # Copy the original matrix
    new_matrix = [row.copy() for row in matrix]

    n_old_users = len(new_matrix)
    n_old_items = len(new_matrix[0])

    # Add 10 new items
    for row in new_matrix:
        row.extend([None] * 10)

    # Give each new item 0–2 ratings from old users
    for item_index in range(n_old_items, n_old_items + 10):
        n_ratings = random.randint(0, 2)
        users = random.sample(range(n_old_users), n_ratings)
        ratings = generate_ratings(n_ratings, dist)

        for user_index, rating in zip(users, ratings):
            new_matrix[user_index][item_index] = rating

    # Add 500 new users
    for _ in range(500):
        new_user = [None] * (n_old_items + 10)

        # Each new user rates 1–3 OLD items
        n_ratings = random.randint(1, 3)
        items = random.sample(range(n_old_items), n_ratings)
        ratings = generate_ratings(n_ratings, dist)

        for item_index, rating in zip(items, ratings):
            new_user[item_index] = rating

        new_matrix.append(new_user)

    return new_matrix
```

```python

def calculate_sparsity(matrix):
    total_cells = len(matrix) * len(matrix[0])
    missing_cells = sum(
        rating is None
        for row in matrix
        for rating in row
    )
    return missing_cells / total_cells * 100


print("Sparsity before:",
      round(calculate_sparsity(uniform_matrix), 2), "%")

print("Sparsity after:",
      round(calculate_sparsity(cold_start_matrix), 2), "%")
```

```python

def count_common_ratings(user_a, user_b):
    return sum(
        a is not None and b is not None
        for a, b in zip(user_a, user_b)
    )


N_OLD_USERS = len(uniform_matrix)

cold_old_overlaps = []

for cold_user in cold_start_matrix[N_OLD_USERS:]:
    for old_user in cold_start_matrix[:N_OLD_USERS]:
        cold_old_overlaps.append(
            count_common_ratings(cold_user, old_user)
        )

no_overlap = sum(x == 0 for x in cold_old_overlaps)

print("COLD-vs-OLD")
print("Number of pairs:", len(cold_old_overlaps))
print("Average co-rated items:",
      sum(cold_old_overlaps) / len(cold_old_overlaps))
print("No-overlap pairs:", no_overlap)
print("No-overlap percentage:",
      no_overlap / len(cold_old_overlaps) * 100, "%")
```

It primarily addresses the **new-user cold-start problem** by collecting initial preferences to help recommend relevant titles to users with little or no rating history.

**6. Why may the global mean and the unweighted mean of user means differ? Are they different in your three matrices?**  
The global mean weights every observed rating equally, while the unweighted mean of user means gives each user equal weight, regardless of how many ratings they provided. Yes, they differ slightly: uniform (3.0038 vs. 3.0040), normal (2.9987 vs. 2.9995), and Poisson (2.9229 vs. 2.9206).

Hiding low ratings based on dislike introduces a systematic positive bias because the remaining visible ratings tend to be higher. Random error generally increases as more ratings are hidden, because fewer observations remain available to estimate the means.
